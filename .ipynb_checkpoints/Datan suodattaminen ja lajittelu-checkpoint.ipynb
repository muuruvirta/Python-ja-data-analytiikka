Tässä demossa käydään läpi datan lajittelu ja suodattaminen Pandasin dataframessa. Käytämme seaborn-kirjastosta löytyvää Titanic dataa. Ladataan ensin data ja tulostetaan siitä 15 ensimmäistä riviä. Huomaa, että sarakkeessa survived: 0 = ei selvinnyt hengissä, 1 = selvisi hengissä. Sarake ei kuitenkaan ole boolean-tyyppinen, vaan sen tietotyyppi on int64. 

Jupyterissa ei kannata käyttää print()-komentoa, koska Jupyter tunnistaa dataframen ja osaa esittää sen html-taulukkona. 

```python
import seaborn as sns

df = sns.load_dataset("titanic")

df.head(15)
```

Suodatetaan seuraavaksi datasta kaikki yli 60-vuotiaan matkustajat. 

```python
df[df["age"]>60]
```

Suodatetaan seuraavaksi kaikki 1. luokan matskustajat. 

```python
df[df["pclass"]==1]
```

Muokataan edellistä suodatusta siten, että suodatetaan kaikki ne 1. luokan matkustajat, jotka selvisivät hengissä. Kiinnitä huomiota sulkujen käyttöön. Niissä menee helposti sekaisin. 

```python
df[
      (df["pclass"]==1) & 
      (df["survived"]==1)
      ]
```

Lajitellaan seuraavaksi aineisto ensin matkalipun hinnan mukaan laskevaan järjestykseen. 

```python
df.sort_values("fare", ascending=False)
```

Huomataan, että useilla matkustajille on samanhintaisia lippuja. Mikäli lipun hinta on sama, niin lajitellaan
matkustajat tällöin iän mukaan nousevaan järjestykseen. 

```python
df.sort_values(
    by=["fare", "age"],
    ascending=[False, True]
)
```

Luodaan seuraavaksi uusi dataframe, johon otetaan mukaan ainoastaan sarakkeet survided, pcclass, age ja fare.

```python
df2 = df[
["survived", "pclass", "age", "fare"]
]

df2
```

Piirretään lopuksi koko aineistosta muutama kaavio. Aloitetaan piirtämällä pylväskaavio selviytyneistä. Korvataan kaaviossa arvot 
0 ja 1 selväkielisellä tekstillä. Käytetään pelkästään Pandasin omia toimintoja. Muita kirjastoja ei siis tarvita. 

```python
df["survived"].replace({0: "No", 1: "Yes"}).value_counts().plot(kind="bar")
```

Seuraavaksi piirretään histogrammin matkustajien ikäjakaumasta. Muokataan sekä kaavion että molempien akselien otsikoita. Käytetään edellisestä demosta tuttua kirjastoa. 

```python
import matplotlib.pyplot as plt

df["age"].hist(bins=20)

plt.title("Matkustajien ikäjakauma")
plt.xlabel("Ikä")
plt.ylabel("Lukumäärä")
plt.show()
```

Tutkitaan seuraavaksi hengissä selviytymisen riippuvuutta matkustusluokasta. Käytetään tähän seaborn-kirjaston barplot-kaaviota. 
Musta viiva on virhepalkki, joka kuvaa arvioon liittyvää epävarmuutta. Tästä puhutaan lisää myöhemmin opintojaksolla. 

```python
import seaborn as sns

sns.barplot(
    data=df,
    x="class",
    y="survived"
)

plt.title("Survival rate by Passenger class")
plt.show()
```

Piirretään lopuksi hajontakaavio, jossa käytetään muuttujina lipun hintaa ja ikää. Tämä kaavio johdattelee korrelaation käsitteeseen, josta puhutaan myöhemmin lisää. 

```python
sns.scatterplot(
    data=df,
    x="age",
    y="fare"
)

plt.title("Ticket price vs. Passenger age")
plt.xlabel("Age")
plt.ylabel("Price")
plt.show()
```
