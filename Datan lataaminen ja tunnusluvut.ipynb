```python
#Otetaan yfinance kirjasto käyttöön
import yfinance as yf

#Sijoitetaan data dataframeen df. Datan aloituspäivä on 1.1.2020.
df = yf.download('^OMXH25', start = '2020-1-1')
```

```python
#Tutkitaan miten data tallentui dataframeen. 
print(df.columns)
```

```python
#Data tallentui ns. multi index -muodossa. Tämä vaikeuttaa analysointia. Poistetaan Ticker-taso. 

df.columns = df.columns.droplevel(1)
```

Lasketaan seuraavaksi datasta yleisimpiä tunnuslukuja. Helpoin tapa on käyttää komentoa df.descibe(), jolloin saadaan kaikki kerralla. Mikäli datamäärä on kuitenkin iso, on monesti tarpeen laskea halutut tunnusluvut yksi kerrallaan. Myöskään mediaania ei tällä komennolla saada. 

```python
df.describe()
```

Lasketaan seuraavaksi päätöskurssin mediaani sekä keskihajonta sekä tulostetaan ne. Sarakkeeseen viitataan hakasuluilla. 
Pyöristetään tulokset kahden desimaalin tarkkuuteen. 

```python
print(f"Mediaani: {df['Close'].median():.2f}")
print(f"Keskihajonta: {df['Close'].std():.2f}")
```

Lisätään seuraavaksi dataframeen yksi uusi sarake, joka kertoo päivän päätös- ja aloituskurssin välisen eron. Tutkitaan onnistuiko se. 

```python
df['Change'] = df['Close']-df['Open']

df.head()
```

Nyt näemme absoluuttisen muutoksen, mutta yleensä prosentuaalinen muutos on tärkeämpi. Lisätään myös sellainen sarake. 

```python
df['Change %'] = (df['Close'] - df['Open']) / df['Open'] * 100

df.head()
```

Siirretään seuraavaksi Volume-sarake viimeiseksi. Poistetaan se ensin ja sijoitetaan sitten uudelleen viimeiseksi. Tämä on 
helpoin ja nopein tapa. 

```python
volume = df.pop('Volume')
df['Volume'] = volume

df.head()
```

Piirretään lopuksi datasta pylväskaavio viimeisen päivän arvoista. Käytetään piirtämiseen
matbloblib.pyplot -kirjastoa. 

```python
import matplotlib.pyplot as plt

viimeinen = df.iloc[-1][['Open', 'High', 'Low', 'Close']] #iloc[-1] valitsee viimeisen rivin

viimeinen.plot(kind='bar') #Tässä annetaan kaavion tyyppi. Muita mahdollisuuksia ovat mm. barh, line, hist, pie, scatter, box ja kde.
#Voit kokeilla niitä, mutta kaikissa ei ole järkeä tämän datan kanssa. 

plt.title('Viimeisimmän päivän kurssitiedot') #x-akselin otsikko
plt.ylabel('Indeksin arvo') #y-akselin otsikko
plt.show() #tulostetaan kaavio 
```

Muutetaan vielä low-palkin väriksi keltainen ja high-palkin väriksi musta. Annetaan kahden reunapalkin väriksi oletussininen
heksadesimaaleina. Kuten tuloksesta huomataan, niin kannattaa käyttää kaavioissa heksavärejä. 

```python
viimeinen.plot(
kind='bar',
color=['#1f77b4', 'yellow', 'black', '#1f77b4']
)

plt.show()
```
