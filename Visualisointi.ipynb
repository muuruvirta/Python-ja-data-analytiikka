# Demo 5: Visualisointi ja muuttujien väliset suhteet

Tässä demossa harjoitellaan aineiston visualisointia Seaborn- ja Matplotlib-kirjastoilla. Tehtävissä 1–9 käytetään Seaborn-kirjaston **penguins-aineistoa**, joka sisältää tietoja pingviinien lajista, painosta, räpylän pituudesta ja muista ominaisuuksista. Tehtävässä 10 käytetään **flights-aineistoa**, joka sisältää lentomatkustajien kuukausittaiset määrät eri vuosilta.

Ensimmäisissä tehtävissä harjoitellaan myös kuvaajan otsikon, akselien otsikoiden, värien, selitteen ja koon muokkaamista. Myöhemmissä tehtävissä tarkastellaan muuttujien välisiä yhteyksiä, regressiosuoraa ja korrelaatiota.

Suorita solut järjestyksessä. Jupyterissa kuvaaja tulostuu suoraan solun alle, joten `plt.show()`-komentoa ei yleensä tarvita, mutta sitä käytetään tässä demossa kuvaajien esittämisen selkeyttämiseksi.

## Aineiston lataaminen

Ladataan tarvittavat kirjastot ja Seaborn-kirjaston `penguins`-aineisto. Tulostetaan aineiston viisi ensimmäistä riviä ja tarkistetaan sarakkeet.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins")

penguins.head()
```

## Tehtävä 1: Histogrammi ja kuvaajan perusmuotoilu

Tee histogrammi pingviinien painosta (`body_mass_g`). Lisää kuvaajalle otsikko sekä x- ja y-akselien otsikot. Muuta histogrammin väriä ja kokeile kuvaajan koon muuttamista.

```python
plt.figure(figsize=(9, 5))

sns.histplot(data=penguins, x="body_mass_g", color="steelblue", bins=20)

plt.title("Pingviinien painojakauma")
plt.xlabel("Paino (g)")
plt.ylabel("Pingviinien lukumäärä")
plt.show()
```

## Tehtävä 2: Jakaumien vertailu

Visualisoi eri pingviinilajien painojakaumat samaan kuvaajaan. Erottele lajit väreillä. Lisää kuvaajalle otsikko, akseleiden otsikot ja selite. Käytä valmista väripalettia.

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=penguins,
    x="body_mass_g",
    hue="species",
    palette="Set2",
    bins=20,
    alpha=0.6
)

plt.title("Pingviinien painojakaumat lajeittain")
plt.xlabel("Paino (g)")
plt.ylabel("Pingviinien lukumäärä")
plt.show()
```

## Tehtävä 3: Lukumäärien visualisointi

Tee pylväsdiagrammi, josta nähdään pingviinien lukumäärä lajeittain.

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=penguins, x="species", hue="species", palette="Set2", legend=False)

plt.title("Pingviinien lukumäärä lajeittain")
plt.xlabel("Laji")
plt.ylabel("Lukumäärä")
plt.show()
```

## Tehtävä 4: Tunnusluvun visualisointi

Selvitä pingviinien keskimääräinen paino lajeittain ja esitä tulos pylväsdiagrammina.

```python
mean_mass = (
    penguins.groupby("species", observed=True)["body_mass_g"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=mean_mass,
    x="species",
    y="body_mass_g",
    hue="species",
    palette="Set2",
    legend=False
)

plt.title("Pingviinien keskimääräinen paino lajeittain")
plt.xlabel("Laji")
plt.ylabel("Keskimääräinen paino (g)")
plt.show()
```

## Tehtävä 5: Jakaumien vertailu laatikkokaaviolla

Tee laatikkokaavio pingviinien painosta lajeittain. Tutki, miten lajien jakaumat eroavat toisistaan ja näkyykö aineistossa poikkeavia havaintoja.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=penguins,
    x="species",
    y="body_mass_g",
    hue="species",
    palette="Set2",
    legend=False
)

plt.title("Pingviinien painojakaumat lajeittain")
plt.xlabel("Laji")
plt.ylabel("Paino (g)")
plt.show()
```

### Laatikkokaavion tulkinta

Laatikkokaavio eli boxplot kuvaa aineiston jakaumaa tiiviissä muodossa.

- **Laatikon keskellä oleva viiva** kuvaa mediaania eli aineiston keskimmäistä arvoa.
- **Laatikon alareuna** kuvaa ensimmäistä kvartiilia (Q1). Noin 25 % havainnoista on tämän arvon alapuolella.
- **Laatikon yläreuna** kuvaa kolmatta kvartiilia (Q3). Noin 75 % havainnoista on tämän arvon alapuolella.
- **Laatikon sisällä** on aineiston keskimmäiset 50 % havainnoista.
- **Laatikosta lähtevät janat eli viikset** kuvaavat aineiston vaihtelua laatikon ulkopuolella. Ne ulottuvat yleensä arvoihin, jotka ovat enintään 1,5 kertaa kvartiilivälin päässä laatikosta.
- **Viiksien ulkopuolella olevat yksittäiset pisteet** ovat mahdollisia poikkeavia havaintoja.

## Tehtävä 6: Räpylän pituuden vertailu

Tee laatikkokaavio räpylän pituudesta (`flipper_length_mm`) lajeittain. Vertaile lajien välisiä eroja.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=penguins,
    x="species",
    y="flipper_length_mm",
    hue="species",
    palette="Set2",
    legend=False
)

plt.title("Pingviinien räpylän pituus lajeittain")
plt.xlabel("Laji")
plt.ylabel("Räpylän pituus (mm)")
plt.show()
```

## Tehtävä 7: Muuttujien välinen yhteys ja regressiosuora

Tee hajontakaavio räpylän pituuden ja painon välisestä suhteesta. Lisää kuvaajaan regressiosuora. Selvitä lisäksi suoran yhtälö muodossa $y = ax + b$ sekä muuttujien välinen Pearsonin korrelaatiokerroin.

```python
regression_data = penguins[["flipper_length_mm", "body_mass_g"]].dropna()

x = regression_data["flipper_length_mm"]
y = regression_data["body_mass_g"]

# np.polyfit palauttaa suoran kulmakertoimen a ja vakiotermin b.
a, b = np.polyfit(x, y, 1)
correlation = x.corr(y)

plt.figure(figsize=(9, 6))

sns.regplot(
    data=regression_data,
    x="flipper_length_mm",
    y="body_mass_g",
    scatter_kws={"alpha": 0.65},
    line_kws={"color": "red"}
)

plt.title("Räpylän pituuden ja painon välinen yhteys")
plt.xlabel("Räpylän pituus (mm)")
plt.ylabel("Paino (g)")
plt.show()

print(f"Suoran yhtälö: y = {a:.2f}x + {b:.2f}")
print(f"Pearsonin korrelaatiokerroin: r = {correlation:.3f}")
```

## Tehtävä 8: Kolmannen muuttujan lisääminen visualisointiin

Tee räpylän pituuden ja painon hajontakaavio siten, että pingviinilajit erotetaan toisistaan väreillä. Vertaa kuvaajaa tehtävän 7 kuvaajaan.

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    palette="Set2",
    s=70,
    alpha=0.75
)

plt.title("Räpylän pituus ja paino lajeittain")
plt.xlabel("Räpylän pituus (mm)")
plt.ylabel("Paino (g)")
plt.legend(title="Laji")
plt.show()
```

## Tehtävä 9: Korrelaatiomatriisi

Laske aineiston numeeristen muuttujien korrelaatiomatriisi ja visualisoi matriisi heatmap-kuvaajana. Selvitä kuvaajan avulla, millä muuttujilla on voimakkain positiivinen ja negatiivinen yhteys.

Tässä vaiheessa korrelaatiota tarkastellaan kuvailevasti. Korrelaation tilastollinen merkitsevyys käsitellään myöhemmin tilastollisen testauksen osiossa.

```python
numeric_data = penguins.select_dtypes(include="number")
correlation_matrix = numeric_data.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    fmt=".2f"
)

plt.title("Pingviiniaineiston korrelaatiomatriisi")
plt.show()
```

## Tehtävä 10: Aikasarjan visualisointi ja kuvaajan tallentaminen

Lataa Seaborn-kirjaston `flights`-aineisto. Aineisto sisältää lentomatkustajien kuukausittaiset määrät vuosina 1949–1960. Muodosta vuoden ja kuukauden perusteella aikamuuttuja ja tee viivakaavio matkustajamäärän kehittymisestä ajan suhteen.

Viimeistele kuvaaja otsikoilla ja sopivalla koolla. Tallenna valmis kuvaaja PNG-tiedostoksi nimellä `lentomatkustajien_maara.png`.

```python
flights = sns.load_dataset("flights")

# Muodostetaan vuoden ja kuukauden perusteella päivämäärä.
flights["date"] = pd.to_datetime(
    flights["year"].astype(str) + "-" + flights["month"].astype(str),
    format="%Y-%b"
)

plt.figure(figsize=(12, 6))

sns.lineplot(data=flights, x="date", y="passengers")

plt.title("Lentomatkustajien määrän kehitys")
plt.xlabel("Aika")
plt.ylabel("Matkustajien määrä")
plt.tight_layout()
plt.savefig("lentomatkustajien_maara.png", dpi=300, bbox_inches="tight")
plt.show()
```
